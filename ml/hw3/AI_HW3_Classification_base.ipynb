## **Теоретическая сноска: почему LogLoss'а так много**

Почти наверное (в математике это значит, во всех случаях, кроме множества размером 0) логлосс набил оскомину за несколько заданий. Давайте посмотрим на него ещё раз:

В логистической регрессии функция потерь

$$\text{LogLoss}(y, \hat{p}) = -\left(y\log \hat{p} + (1-y)\log (1-\hat{p})\right)$$

Зачем мы так долго с ней возимся?

#### **Пункт 1.**
Во-первых, это **следствие максимизации правдоподобия** при биномиальной модели.

Если считать, что целевая переменная (Y\in{0,1}) распределена как

$$P(Y=1 \mid x) = \hat{p}(x), \qquad P(Y=0\mid x) = 1-\hat{p}(x),$$
то правдоподобие выборки ( (x_i, y_i) )\ равно
$$L = \prod_{i=1}^n \hat{p}_i^{y_i}(1-\hat{p}_i)^{1-y_i}.$$

Максимизация $\log L$ эквивалентна минимизации LogLoss.
Таким образом, LogLoss — **единственная функция потерь, полностью согласованная с вероятностной моделью логистической регрессии**.

#### **Пункт 2.**

Во-вторых, логлосс поможет нам в будущем понять другие функции потерь. Так, например LogLoss является частным случаем **кросс-энтропии между истинным распределением и предсказанным**.

Для двух распределений $p$ (истинного) и $q$ (предсказанного) кросс-энтропия определяется как

$$H(p,q) = -\sum_{k} p(k)\log q(k).$$

В бинарном случае истинное распределение дискретно:

$$p = (y, 1-y), \qquad q = (\hat{p}, 1-\hat{p}),$$
и подстановка даёт

$$H(p,q) = -\left[y\log \hat{p} + (1-y)\log (1-\hat{p})\right] = \text{LogLoss}.$$

**Шаблон класса описан ниже, вам нужно реализовать каждую из заготовленных функций.**

**ВАЖНО!** Мы заполняем данный шаблон, даже если он нам не нравится. Менять структуру класса и писать по-своему запрещено - за это будут сняты баллы.

```python
import numpy as np
from sklearn.base import BaseEstimator

class LogReg(BaseEstimator):
    def __init__(self, gd_type='stochastic',
                 tolerance=1e-4, max_iter=1000, w0=None, eta=1e-2):
        """
        gd_type: 'full' or 'stochastic'
        tolerance: for stopping gradient descent
        max_iter: maximum number of steps in gradient descent
        w0: np.array of shape (d) — init weights
        eta: learning rate
        """
        self.gd_type = gd_type
        self.tolerance = tolerance
        self.max_iter = max_iter
        self.w0 = w0
        self.w = None
        self.eta = eta
        self.loss_history = []

    def fit(self, X, y):
        """
        X: np.array of shape (ell, d)
        y: np.array of shape (ell)
        ---
        output: self
        """
        # Добавляем константный признак для bias
        X_with_bias = np.hstack([np.ones((X.shape[0], 1)), X])
        
        # Инициализация весов
        if self.w0 is None:
            self.w = np.zeros(X_with_bias.shape[1])
        else:
            # Если w0 задан, добавляем bias как 0
            if len(self.w0) == X.shape[1]:
                self.w = np.hstack([np.array([0.0]), self.w0])
            else:
                self.w = np.zeros(X_with_bias.shape[1])
        
        self.loss_history = []
        
        # Вычисляем начальный loss
        initial_loss = self.calc_loss(X_with_bias, y)
        self.loss_history.append(initial_loss)
        
        # Градиентный спуск
        for iteration in range(self.max_iter):
            w_old = self.w.copy()
            
            if self.gd_type == 'full':
                gradient = self.calc_gradient(X_with_bias, y)
                self.w = self.w - self.eta * gradient
            elif self.gd_type == 'stochastic':
                idx = np.random.randint(0, X_with_bias.shape[0])
                X_sample = X_with_bias[idx:idx+1]
                y_sample = y[idx:idx+1]
                gradient = self.calc_gradient(X_sample, y_sample)
                self.w = self.w - self.eta * gradient
            
            # Вычисляем loss после обновления весов
            loss = self.calc_loss(X_with_bias, y)
            self.loss_history.append(loss)
            
            # Проверка критерия останова
            if np.linalg.norm(self.w - w_old) < self.tolerance:
                break
        
        return self

    def sigmoid(self, x):
        # Защита от переполнения
        x_clipped = np.clip(x, -500, 500)
        return 1 / (1 + np.exp(-x_clipped))

    def predict_proba(self, X):
        if self.w is None:
            raise Exception('Not trained yet')
        
        # Добавляем константный признак
        X = np.hstack([np.ones((X.shape[0], 1)), X])
        return self.sigmoid(X @ self.w)

    def predict(self, X):
        if self.w is None:
            raise Exception('Not trained yet')
        
        proba = self.predict_proba(X)
        return (proba >= 0.5).astype(int)

    def calc_gradient(self, X, y):
        """
        X: np.array of shape (ell, d) (ell can be equal to 1 if stochastic)
        y: np.array of shape (ell)
        ---
        output: np.array of shape (d)
        """
        # В качестве лосса тут рассматриваем LogLoss !
        p = self.sigmoid(X @ self.w)
        # Градиент LogLoss: mean((p - y) * X)
        gradient = np.mean((p - y).reshape(-1, 1) * X, axis=0)
        return gradient

    def calc_loss(self, X, y):
        """
        X: np.array of shape (ell, d)
        y: np.array of shape (ell)
        ---
        output: float
        """
        p = self.sigmoid(X @ self.w)
        # Защита от log(0)
        epsilon = 1e-15
        p = np.clip(p, epsilon, 1 - epsilon)
        # считаем по фрмуле LogLoss: -mean(y * log(p) + (1-y) * log(1-p))
        loss = -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))
        return loss
```

### **Задание 12. Улучшение базовых моделей за счет данных. (0.3 балла).**

- Подберите гиперпараметры CountVectorizer так, чтобы признаков было минимум в 4 раза меньше, чем объектов, а качество модели при этом изменилось не более чем на $\pm 0.07$. Опишите подобранные гиперпараметры и на что они влияют.

Обучайте и логистическую регрессию, и SVC.

```python
n_samples = len(X_train)
max_features_target = n_samples // 4
print(f"Количество объектов: {n_samples}")
print(f"Максимальное количество признаков (n_samples/4): {max_features_target}")

baseline_auc_lr = roc_auc_score(y_test, y_proba_lr)
baseline_auc_svc = roc_auc_score(y_test, y_proba_svc)

print(f"Базовое качество:")
print(f"  LR ROC-AUC: {baseline_auc_lr:.4f}")
print(f"  SVC ROC-AUC: {baseline_auc_svc:.4f}")
```

max_features: максимальное количество признаков (слов)
- Ограничивает размерность пространства признаков
- Выбирает наиболее частые слова

min_df: минимальная частота документа
- Игнорирует слова, встречающиеся менее чем в min_df документах
- Помогает убрать редкие слова

max_df: максимальная частота документа
- Игнорирует слова, встречающиеся более чем в max_df*100% документов
- Помогает убрать слишком частые (стоп-слова)

ngram_range: диапазон n-грамм
- (1,1) - только униграммы (отдельные слова)
- (1,2) - униграммы и биграммы (пары слов)

```python
best_params_lr = None
best_params_svc = None
best_auc_lr = -1  # Начинаем с -1, чтобы найти любую комбинацию в пределах ±0.07
best_auc_svc = -1
best_vectorizer_lr = None
best_vectorizer_svc = None

# Расширяем диапазон параметров для более тщательного поиска
# Убеждаемся, что max_features не превышает max_features_target
max_features_options = [500, 1000, 1500, 2000]
max_features_options = [f for f in max_features_options if f <= max_features_target]
if max_features_target not in max_features_options:
    max_features_options.append(max_features_target)
    
min_df_options = [1, 2, 3, 5]  # минимальная частота документа
max_df_options = [0.5, 0.7, 0.8, 0.9, 1.0]  # максимальная частота документа
ngram_range_options = [(1, 1), (1, 2)]  # униграммы или биграммы
```

```python
print("Начинаем подбор гиперпараметров...")
total_combinations = len(max_features_options) * len(min_df_options) * len(max_df_options) * len(ngram_range_options)
print(f"Всего комбинаций для проверки: {total_combinations}")

for idx, max_features in enumerate(max_features_options):
    if max_features > max_features_target:
        continue
    for min_df in min_df_options:
        for max_df in max_df_options:
            for ngram_range in ngram_range_options:
                try:
                    vectorizer = CountVectorizer(
                        max_features=max_features,
                        min_df=min_df,
                        max_df=max_df,
                        ngram_range=ngram_range
                    )

                    X_train_vec = vectorizer.fit_transform(X_train)
                    X_test_vec = vectorizer.transform(X_test)
                    
                    # Проверяем условие на количество признаков
                    n_features = X_train_vec.shape[1]
                    if n_features > max_features_target:
                        continue
                    
                    # Проверяем условие: признаков должно быть минимум в 4 раза меньше объектов
                    if n_features * 4 > X_train_vec.shape[0]:
                        continue
                    
                    # Обучаем LR
                    lr = LogisticRegression(random_state=42, max_iter=1000)
                    lr.fit(X_train_vec, y_train)
                    y_proba_lr_test = lr.predict_proba(X_test_vec)[:, 1]
                    auc_lr = roc_auc_score(y_test, y_proba_lr_test)
                    
                    # Ищем максимум качества в пределах +-0.07 от baseline
                    if abs(auc_lr - baseline_auc_lr) <= 0.07:
                        if best_auc_lr < 0 or auc_lr > best_auc_lr:
                            best_auc_lr = auc_lr
                            best_params_lr = {
                                'max_features': max_features,
                                'min_df': min_df,
                                'max_df': max_df,
                                'ngram_range': ngram_range
                            }
                            best_vectorizer_lr = vectorizer
                            print(f"Найдены лучшие параметры для LR: AUC={auc_lr:.4f}, n_features={n_features}")
                    
                    # Обучаем SVC (только если признаков не слишком много для ускорения)
                    if n_features <= 2000:  # Ограничиваем для SVC из-за времени обучения
                        svc = SVC(random_state=42, probability=True)
                        svc.fit(X_train_vec, y_train)
                        y_proba_svc_test = svc.predict_proba(X_test_vec)[:, 1]
                        auc_svc = roc_auc_score(y_test, y_proba_svc_test)
                        
                        # Ищем максимум качества в пределах ±0.07 от baseline
                        if abs(auc_svc - baseline_auc_svc) <= 0.07:
                            if best_auc_svc < 0 or auc_svc > best_auc_svc:
                                best_auc_svc = auc_svc
                                best_params_svc = {
                                    'max_features': max_features,
                                    'min_df': min_df,
                                    'max_df': max_df,
                                    'ngram_range': ngram_range
                                }
                                best_vectorizer_svc = vectorizer
                                print(f"Найдены лучшие параметры для SVC: AUC={auc_svc:.4f}, n_features={n_features}")
                except Exception as e:
                    # Пропускаем комбинации, которые вызывают ошибки
                    continue

print("Подбор гиперпараметров завершен!")
```

```python
print(f"  LR ROC-AUC: {baseline_auc_lr:.4f}")
print(f"  SVC ROC-AUC: {baseline_auc_svc:.4f}")
```
