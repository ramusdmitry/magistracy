## 1. Анализ исторических данных

### 1.1 Вычисление скользящего среднего и стандартного отклонения


Загрузка данных

````python
import time
from concurrent.futures import ProcessPoolExecutor, ThreadPoolExecutor
import multiprocessing

df = pd.read_csv('temperature_data.csv', parse_dates=['timestamp'])
print(f"Загружено {len(df)} записей")
print(f"Города: {df['city'].unique()}")
print(f"\nОписательная статистика:")
df.describe()
````

````python
def analyze_city_data(city_data):
    """
    Анализ данных для одного города:
    - Скользящее среднее (окно 30 дней)
    - Скользящее стандартное отклонение
    - Определение аномалий
    """
    city_df = city_data.copy()
    city_df = city_df.sort_values('timestamp')
    
    city_df['rolling_mean'] = city_df['temperature'].rolling(window=30, center=True).mean()
    city_df['rolling_std'] = city_df['temperature'].rolling(window=30, center=True).std()
    
    city_df['lower_bound'] = city_df['rolling_mean'] - 2 * city_df['rolling_std']
    city_df['upper_bound'] = city_df['rolling_mean'] + 2 * city_df['rolling_std']
    city_df['is_anomaly'] = (city_df['temperature'] < city_df['lower_bound']) | \
                            (city_df['temperature'] > city_df['upper_bound'])
    
    return city_df
````

````python
# Функция для анализа одного города (для параллельного выполнения)
def process_city(args):
    city, group_data = args
    return analyze_city_data(group_data)
````

````python
print("Анализ данных без распараллеливания...")
start_time = time.time()

results_sequential = []
for city in df['city'].unique():
    city_data = df[df['city'] == city].copy()
    analyzed = analyze_city_data(city_data)
    results_sequential.append(analyzed)

df_analyzed_seq = pd.concat(results_sequential, ignore_index=True)
sequential_time = time.time() - start_time
print(f"Время выполнения (последовательно): {sequential_time:.4f} сек")
print(f"Найдено аномалий: {df_analyzed_seq['is_anomaly'].sum()}")
````

### 1.2 Исследование распараллеливания анализа

**Теория о распараллеливании:**
- **ThreadPoolExecutor** - использует потоки, хорош для I/O-bound задач (сеть, диск)
- **ProcessPoolExecutor** - использует процессы, хорош для CPU-bound задач, но имеет накладные расходы на сериализацию
- В Jupyter Notebook ProcessPoolExecutor имеет ограничения из-за невозможности сериализации функций, определенных в __main__
- Для pandas операций с GIL распараллеливание может не давать значительного ускорения


````python
# Подготовка данных для параллельной обработки
city_groups = [(city, df[df['city'] == city].copy()) for city in df['city'].unique()]

# Тест 1: ThreadPoolExecutor
print("Анализ данных с ThreadPoolExecutor...")
start_time = time.time()

with ThreadPoolExecutor(max_workers=multiprocessing.cpu_count()) as executor:
    results_thread = list(executor.map(process_city, city_groups))

df_analyzed_thread = pd.concat(results_thread, ignore_index=True)
thread_time = time.time() - start_time
print(f"Время выполнения (ThreadPoolExecutor): {thread_time:.4f} сек")
print(f"Найдено аномалий: {df_analyzed_thread['is_anomaly'].sum()}")
````

## 2. Мониторинг текущей температуры через OpenWeatherMap API

### 2.1 Синхронный и асинхронный доступ к API

**Теория:**
- **Синхронный запрос** (requests): блокирует выполнение до получения ответа
- **Асинхронный запрос** (aiohttp): позволяет выполнять несколько запросов параллельно

**Когда что использовать:**
- **Синхронный**: для одиночных запросов, простых скриптов
- **Асинхронный**: для множественных запросов к API, веб-приложений с высокой нагрузкой

В Streamlit предпочтительнее синхронный подход, т.к.:
1. Streamlit сам управляет concurrency
2. Код проще и понятнее
3. Для одиночных запросов разница минимальна


````python
API_KEY = "бро просто введи меня" 
````

````python
import requests
import asyncio
import aiohttp
import nest_asyncio

# Разрешаем вложенные event loops для Jupyter
nest_asyncio.apply()

# Словарь для маппинга городов на английские названия для API
city_names_en = {
    "New York": "New York",
    "London": "London",
    "Paris": "Paris",
    "Tokyo": "Tokyo",
    "Moscow": "Moscow",
    "Sydney": "Sydney",
    "Berlin": "Berlin",
    "Beijing": "Beijing",
    "Rio de Janeiro": "Rio de Janeiro",
    "Dubai": "Dubai",
    "Los Angeles": "Los Angeles",
    "Singapore": "Singapore",
    "Mumbai": "Mumbai",
    "Cairo": "Cairo",
    "Mexico City": "Mexico City"
}

def get_current_temperature_sync(city: str, api_key: str) -> dict:
    """
    Синхронный запрос к OpenWeatherMap API
    Возвращает словарь с температурой или ошибкой
    """
    base_url = "https://api.openweathermap.org/data/2.5/weather"
    params = {
        "q": city,
        "appid": api_key,
        "units": "metric"  # Температура в Цельсиях
    }
    
    try:
        response = requests.get(base_url, params=params, timeout=10)
        data = response.json()
        
        if response.status_code == 200:
            return {
                "success": True,
                "city": city,
                "temperature": data["main"]["temp"],
                "feels_like": data["main"]["feels_like"],
                "description": data["weather"][0]["description"],
                "humidity": data["main"]["humidity"]
            }
        else:
            return {
                "success": False,
                "city": city,
                "error": data.get("message", "Unknown error"),
                "code": data.get("cod")
            }
    except Exception as e:
        return {
            "success": False,
            "city": city,
            "error": str(e)
        }

async def get_current_temperature_async(city: str, api_key: str, session: aiohttp.ClientSession) -> dict:
    """
    Асинхронный запрос к OpenWeatherMap API
    """
    base_url = "https://api.openweathermap.org/data/2.5/weather"
    params = {
        "q": city,
        "appid": api_key,
        "units": "metric"
    }
    
    try:
        async with session.get(base_url, params=params, timeout=aiohttp.ClientTimeout(total=10)) as response:
            data = await response.json()
            
            if response.status == 200:
                return {
                    "success": True,
                    "city": city,
                    "temperature": data["main"]["temp"],
                    "feels_like": data["main"]["feels_like"],
                    "description": data["weather"][0]["description"],
                    "humidity": data["main"]["humidity"]
                }
            else:
                return {
                    "success": False,
                    "city": city,
                    "error": data.get("message", "Unknown error"),
                    "code": data.get("cod")
                }
    except Exception as e:
        return {
            "success": False,
            "city": city,
            "error": str(e)
        }

async def get_temperatures_for_cities_async(cities: list, api_key: str) -> list:
    """
    Получить температуру для нескольких городов асинхронно
    """
    async with aiohttp.ClientSession() as session:
        tasks = [get_current_temperature_async(city, api_key, session) for city in cities]
        results = await asyncio.gather(*tasks)
    return results
````

### 2.3 Проверка аномальности текущей температуры


````python
from datetime import datetime

def get_current_season():
    """Определить текущий сезон по дате"""
    month = datetime.now().month
    return month_to_season[month]

def check_temperature_anomaly(city: str, current_temp: float, historical_data: pd.DataFrame) -> dict:
    """
    Проверяет, является ли текущая температура аномальной
    на основе исторических данных для текущего сезона
    
    Returns:
        dict с информацией об аномальности
    """
    season = get_current_season()
    
    # Фильтруем данные по городу и сезону
    city_season_data = historical_data[
        (historical_data['city'] == city) & 
        (historical_data['season'] == season)
    ]
    
    if len(city_season_data) == 0:
        return {
            "is_anomaly": None,
            "message": f"Нет исторических данных для {city} в сезоне {season}"
        }
    
    # Вычисляем статистику
    mean_temp = city_season_data['temperature'].mean()
    std_temp = city_season_data['temperature'].std()
    lower_bound = mean_temp - 2 * std_temp
    upper_bound = mean_temp + 2 * std_temp
    
    is_anomaly = current_temp < lower_bound or current_temp > upper_bound
    
    if current_temp < lower_bound:
        anomaly_type = "холодная"
    elif current_temp > upper_bound:
        anomaly_type = "жаркая"
    else:
        anomaly_type = None
    
    return {
        "city": city,
        "season": season,
        "current_temp": current_temp,
        "mean_temp": round(mean_temp, 1),
        "std_temp": round(std_temp, 1),
        "lower_bound": round(lower_bound, 1),
        "upper_bound": round(upper_bound, 1),
        "is_anomaly": is_anomaly,
        "anomaly_type": anomaly_type,
        "message": f"{'🔴 АНОМАЛИЯ (' + anomaly_type + ')' if is_anomaly else '🟢 Норма'}: {current_temp:.1f}°C (норма: {lower_bound:.1f}°C - {upper_bound:.1f}°C)"
    }
````

Демонстрация на примере

````python
print(f"Текущий сезон: {get_current_season()}")
print("Пример проверки аномальности (на синтетических данных):")

# Тестируем на исторических данных
for city in ["Moscow", "Berlin", "Cairo"]:
    # Симулируем "текущую" температуру на основе среднего + случайное отклонение
    city_data = df_analyzed_seq[df_analyzed_seq['city'] == city]
    season_data = city_data[city_data['season'] == get_current_season()]
    simulated_temp = season_data['temperature'].mean() + np.random.normal(0, 3)
    
    result = check_temperature_anomaly(city, simulated_temp, df_analyzed_seq)
    print(f"\n{city}: {result['message']}")
````

### 4.2 О синхронном и асинхронном доступе к API

| Подход | Когда использовать |
|--------|-------------------|
| **Синхронный (requests)** | Одиночные запросы, простые приложения, Streamlit |
| **Асинхронный (aiohttp)** | Множественные запросы, высоконагруженные системы |

**Вывод для Streamlit**: Используем синхронный подход, т.к.:
1. Streamlit сам управляет concurrency и кэшированием
2. Для одиночных запросов разница минимальна
3. У нас нет массовых запросов на наш бекенд
4. Код проще в поддержке и отладке
   
- Асинхронный подход был бы оправдан при:
  1. Массовых запросах (десятки/сотни городов)
  2. Высоконагруженных приложениях
  3. Когда важна минимальная латентность при параллельных операциях

### 4.3 Об аномалиях

- Аномалии определяются как температуры, выходящие за пределы μ ± 2σ
- Для каждого города ~3-5% дней являются аномальными
- Аномалии распределены равномерно по сезонам
